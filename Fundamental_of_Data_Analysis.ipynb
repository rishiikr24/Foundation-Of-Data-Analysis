```python
9 import os

STRUCTURED_EXTENSIONS = {'.csv','.xls','.xlsx','.db'}
SEMI_STRUCTURED_EXTENSIONS = {'.json','.xml','.yaml','.yml'}
UNSTRUCTURED_EXTENSIONS = {'.txt','.pdf','.jpg','.jpeg','.png','.mp4','.avi','.doc','.docx'}

def identify_file_type(file_path):
  ext = os.path.splitext(file_path)[1] # Access the second element of the tuple
  ext = ext.lower()

  if ext in STRUCTURED_EXTENSIONS:
        return "Structured"
  elif ext in SEMI_STRUCTURED_EXTENSIONS :
        return "Semi-Structured"
  elif ext in UNSTRUCTURED_EXTENSIONS:
        return "Unstructured"
  else:
        return "Unknown"

files={
    "data.csv",
    "config.json",
    "document.pdf",
    "image.jpg",
    "records.xlsx",
    "video.mp4",
    "note.txt"
}

for f in files:
    print(f"{f} --> {identify_file_type(f)}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

col = 'petal length (cm)'

plt.figure(figsize=(8, 5))
sns.histplot(df[col], kde=True, bins=20, color="skyblue", edgecolor="black")
plt.xlabel(col)
plt.ylabel("Frequency")
plt.show()

skewness = skew(df[col])
kurt = kurtosis(df[col])

print(f"Skewness of {col}: {skewness:.3f}")
print(f"Kurtosis of {col}: {kurt:.3f}")

if skewness > 0:
    print("The distribution is positively skewed (longer right tail).")
elif skewness < 0:
    print("The distribution is negatively skewed (longer left tail).")
else:
    print("The distribution is symmetric.")

if kurt > 0:
    print("The distribution is leptokurtic (sharper peak, heavier tail than normal).")
elif kurt < 0:
    print("The distribution is platykurtic (flatter peak, lighter tail than normal).")
else:
    print("The distribution is mesokurtic.")
```

# Task
Combine student grades with attendance and demographic data using merge() and join().

## Create sample dataframes

### Subtask:
Create three sample pandas DataFrames for student grades, attendance, and demographic data.


**Reasoning**:
Create three sample pandas DataFrames for student grades, attendance, and demographic data with specified columns and sample data.



```python
import pandas as pd

# Create grades_df
grades_data = {
    'StudentID': [1, 1, 2, 2, 3, 3],
    'Subject': ['Math', 'Science', 'Math', 'Science', 'Math', 'Science'],
    'Grade': ['A', 'B', 'C', 'A', 'B', 'C']
}
grades_df = pd.DataFrame(grades_data)

# Create attendance_df
attendance_data = {
    'StudentID': [1, 1, 2, 3, 3, 3],
    'Date': ['2023-01-01', '2023-01-02', '2023-01-01', '2023-01-01', '2023-01-02', '2023-01-03'],
    'AttendanceStatus': ['Present', 'Present', 'Absent', 'Present', 'Absent', 'Present']
}
attendance_df = pd.DataFrame(attendance_data)

# Create demographics_df
demographics_data = {
    'StudentID': [1, 2, 3],
    'Age': [16, 17, 16],
    'Gender': ['Male', 'Female', 'Male']
}
demographics_df = pd.DataFrame(demographics_data)

print("Original DataFrames:")
print("Grades DataFrame:")
display(grades_df)
print("\nAttendance DataFrame:")
display(attendance_df)
print("\nDemographics DataFrame:")
display(demographics_df)

# Merge DataFrames
merged_grades_attendance = pd.merge(grades_df, attendance_df, on='StudentID')
merged_all = pd.merge(merged_grades_attendance, demographics_df, on='StudentID')

print("\nMerged DataFrame:")
display(merged_all)


# Join DataFrames
grades_df_indexed = grades_df.set_index('StudentID')
attendance_df_indexed = attendance_df.set_index('StudentID')
demographics_df_indexed = demographics_df.set_index('StudentID')

joined_grades_attendance = grades_df_indexed.join(attendance_df_indexed, how='left')
joined_all = joined_grades_attendance.join(demographics_df_indexed, how='left')

print("\nJoined DataFrame:")
display(joined_all)
```

## Summary:

### Data Analysis Key Findings

* The `merge()` function successfully combined the three DataFrames (`grades_df`, `attendance_df`, and `demographics_df`) into a single DataFrame (`merged_all`) using the 'StudentID' column as the key.
* The `join()` function also successfully combined the three DataFrames into a single DataFrame (`joined_all`) after setting 'StudentID' as the index for each individual DataFrame.
* Both `merge()` and `join()` can be used to combine DataFrames, but `join()` requires setting the joining column(s) as the index.

### Insights or Next Steps

* The choice between `merge()` and `join()` depends on whether the joining column is already the index or if it's more convenient to set it as the index.
* Further analysis can be performed on the combined DataFrames to explore relationships between grades, attendance, and demographics.

```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
x = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([40, 50, 60, 65, 70, 80])
model = LinearRegression()
model.fit(x, y)
y_pred = model.predict(x)
slope = model.coef_[0]
intercept = model.intercept_
r2 = r2_score(y, y_pred)
print("Slope (m):", slope)
print("Intercept (c):", intercept)
print("R-squared value:", r2)
plt.scatter(x, y, color='blue', label="Actual Data")
plt.plot(x, y_pred, color='red', label="Regression Line")
plt.title("Simple Linear Regression: Hours vs Exam Score")
plt.xlabel("Hours Studied (X)")
plt.ylabel("Exam Score (Y)")
plt.legend()
plt.grid(True)
plt.show()
```
